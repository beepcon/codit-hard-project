```python
import json
from datetime import datetime
from importlib.metadata import PackageNotFoundError, distribution
from pathlib import Path

import pandas as pd

names = [l.strip() for l in Path('../requirements.txt').read_text(encoding='utf-8').splitlines()
         if l.strip() and not l.startswith(('#', '-'))]
# requirements에는 없지만 00 설치 셀이 함께 올리는 패키지
names += [n for n in ['huggingface_hub', 'beautifulsoup4', 'torchaudio'] if n not in names]
rows, pins = [], []
for n in names:
    try:
        d = distribution(n)
    except PackageNotFoundError:
        rows.append({'package': n, 'version': '미설치', 'installed_at': '', 'source': ''})
        continue
    url = d.read_text('direct_url.json')
    src = json.loads(url) if url else {}
    commit = src.get('vcs_info', {}).get('commit_id', '')
    rows.append({'package': n, 'version': d.version,
                 'installed_at': f'{datetime.fromtimestamp(Path(d._path).stat().st_mtime):%m-%d %H:%M}',
                 'source': f"git {commit[:8]}" if commit else 'PyPI'})
    pins.append(f"{n} @ git+{src['url']}@{commit}" if commit else f'{n}=={d.version}')
display(pd.DataFrame(rows))
print('\n'.join(pins))
```
